TYPES OF PROMPT ENGINEEERING :-
     * Zero shot - providing a promt without examples for the model to follow.
     * One shot - Including a single example demonstarion in the prompt.
     * Few shot - Providing a few example demonstrations for the model.
     * Chain of Thought - Guiding the model through a reasoning process step by step.
