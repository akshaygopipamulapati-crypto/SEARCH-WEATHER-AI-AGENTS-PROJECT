Temperature 
   temperature controls how deterministic or random the output is 
      > Lower values produce more predictable response 
      > higher values allows for more variation
      
TEMPERATURE     USE

  0.0            Very deterministic
  0.2            factual Q&A
  0.5            Balanced & focused
  0.7            general chatbot
  1.0            Creative 
  1.5            very creative
  2.0            maximum allowed by groq API

2 Max Output 
   > Max output sets the limit of how many tokens the LLM can generate in a response. 
   > Tokens are not exactly words they can be whole words or fragments depending on language model. 
   > this parameter helps to control length of generated text.

```python
import os

from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model(
    'openai/gpt-oss-120b',
     model_provider='groq',
     temperature = 0.2,
     max_tokens = 150)

response = model.invoke("What is your Name?")

print(response.content)
```

3. TOP_P SAMPLING
    
    * top_p is called nuclear sampling.
    * instead of considering every possible next token , the model considers tokens whose combined probability reaches p.

TOP_P          Behavior

 0.1             Very Focused
 0.5             Focused
 0.9             Diverse
 1.0             Max probability range
 

```python
import os

from langchain.chat_models import init_chat_model

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model(
    'openai/gpt-oss-120b',
     model_provider='groq',
     temperature = 0.2,
     max_tokens = 2500,
     model_kwargs={
        "top_p":0.1
     })

response = model.invoke("Tell me about  Akshay Gopi Pamulapati studying in QIS college in ongole")

print(response.content)
```

BATCH - 
 
     * Batch is a collection of independent requests to a model can significantly improve performance and reduce costs,as the processing can be done in parellel.

```python
responses = model.batch([
    "what is Generative AI?",
    "How do we reach a HR's in Linkedin Easily?",
    "Which is the best peaceful place in the ongole?",
])
for response in responses:
    print(response)
```

```python
responses = model.batch([
    "what is Generative AI?",
    "How do we reach a HR's in Linkedin Easily?",
    "Which is the best peaceful place in the ongole?"],
    config={
        'max_occurency' : 5,
    })
for response in responses:
    print(response.text)
```

TOOLS  

    *  Models can request to call tools that perform tasks sach as fetching data fro a database, searching the web, or running code.
    *  Tools are pairing of:
                  1* A schema, including the name of the tool, a description,and/or argument definitions(often a JSON schema)
                  2* A function or Coroutine to Execute.

```python
from langchain.tools import tool

@tool
def get_weather(location: str) -> str:
    """Get the weather at a location."""
    return f"The weather at {location} is sunny."


result = get_weather.invoke({
    "location": "Addanki"
})

print(result)
```
