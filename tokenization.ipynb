```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
text = "akshay doing errors"
```

```python
tokens = tokenizer.tokenize(text)
```

```python
print("Tokens: ", tokens)
```

```python
# Encode 
encoded = tokenizer(text)
```

```python
print("\nInputs IDs:", encoded["input_ids"])
```

```python
print('length of the tokens',len(tokens))
```

```python
vocab = tokenizer.get_vocab()
```

```python
print("length of the vocabulary", len(vocab))
```

```python
import tiktoken
```

```python
encoding = tiktoken.encoding_for_model('gpt-4o')
```

```python
text = "Lover Boy Asif"
```

```python
tokens = encoding.encode(text)
```

```python
print("Tokens:",tokens)
```

```python
print("\nIndividual tokens:")
for token in tokens:
    print(encoding.decode([token]), end="|")
```
